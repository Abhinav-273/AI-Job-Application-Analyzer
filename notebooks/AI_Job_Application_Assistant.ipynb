```python
#Importing the libraries

from google.colab import files
import pymupdf
import pandas as pd
import numpy as np

uploaded = files.upload()
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print(uploaded.keys())
```

```python
#Updating the PDF extraction

pdf_file = list(uploaded.keys())[0]
doc = pymupdf.open(pdf_file)
resume_text = ""
for page in doc:
  resume_text += page.get_text()

print("Number of characters: ", len(resume_text))
print("\n Resume Preview: \n")
print(resume_text[:3000])
```

```python
#Adding the job Description

job_description = """
Job Title: Associate Machine Learning EngineerJob Overview:
We are seeking an Associate Machine Learning Engineer to design, build, and deploy data-driven predictive models and automated integration pipelines. In this role, you will analyze large datasets, develop machine learning algorithms—ranging from time-series forecasting to deep learning models—and integrate enterprise data workflows using cloud technologies. You will collaborate with cross-functional teams to clean complex data, optimize model performance, and ensure smooth data flow across enterprise applications.  Key Responsibilities:Model Development & Optimization: Design, train, and evaluate machine learning models (such as XGBoost, Random Forest, K-Means, and LSTM-Attention architectures) for classification, regression, and time-series tasks.  Feature Engineering & Data Preprocessing: Perform data cleaning, feature extraction, normalization, and Exploratory Data Analysis (EDA) on large-scale datasets using Python, Pandas, NumPy, Scikit-learn, and SQL.  Data Integration & Pipeline Automation: Build, test, and maintain secure automated ETL pipelines and SFTP-based workflows using cloud tools like Oracle Integration Cloud (OIC) to move, transform, and validate enterprise data.  Performance Tuning & Evaluation: Fine-tune hyperparameter settings and evaluate model precision using standard regression and classification metrics ($R^2$, MAE, RMSE).  System Monitoring & Troubleshooting: Diagnose workflow bottlenecks, troubleshoot data pipeline failures, and optimize end-to-end performance across cloud environments.  Qualifications & Requirements:Education: Bachelor’s degree in Electronics & Communication Engineering, Computer Science, or a related field.  Technical Skills: Strong proficiency in Python and SQL.  ML Libraries & Tools: Experience with Scikit-learn, Pandas, NumPy, Matplotlib, Seaborn, and time-series/deep learning frameworks.  Cloud & Integration Platforms: Familiarity with Oracle Cloud Infrastructure (OCI), OIC, Oracle VBCS, Oracle APEX, and secure data transfer protocols (SFTP).  Certifications: Professional certifications in AI/ML or Enterprise Cloud Integrations are preferred.
"""

print(job_description)
```

```python
#Resume-job Similarity

documents = [resume_text, job_description]

vectorizer = TfidfVectorizer(
    stop_words = "english"
)
tfidf_matrix = vectorizer.fit_transform(documents)

similarity_score = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]
match_percentage = similarity_score * 100

print(f"Resume-Job Match Score {match_percentage:.2f}%")
```

```python
#LLM Analysis Function

def analyze_resume_with_llm(
    resume_text,
    job_description,
    matched_skills,
    missing_skills,
    skill_match_score):

  prompt = f"""

  You are a AI Career assistant.
  Analyze this resume against the job description.

  RESUME:
  {resume_text}

  JOB DESCRIPTION:
  {job_description}

  MATCHED SKILLS:
  {", ".join(matched_skills)}

  MISSING SKILLS:
  {", ".join(missing_skills)}

  SKILL MATCH SCORE:
  {skill_match_score:.2f}%


  Returnn your response using exactly these five sections:

  Give a short assessment.

  Give 3-5 relevant strength.

  Give the most important missing or weak areas.

  Give 3-5 specific suggestions.

  Give 5 technical interview questions relevant to the job.

  Keep the response concise and practical.
  """

  response = client.models.generate_content(
      model = "gemini-3.6-flash",
      contents = prompt
  )
  return response.text
```

```python
#Call the function

llm_analysis = analyze_resume_with_llm(
    resume_text,
    job_description,
    matched_skills,
    missing_skills,
    skill_match_score
)
print(llm_analysis)
```

```python
#Final Analysis Function

def analyze_resume(resume_text, job_description):
  resume_lower = resume_text.lower()
  job_lower = job_description.lower()

  resume_skills = []
  for skill, aliases in skill_aliases.items():
    for alias in aliases:
      if alias in resume_lower:
        resume_skills.append(skill)
        break
  job_skills = []

  for skill, aliases in skill_aliases.items():
    for alias in aliases:
      if alias in job_lower:
        job_skills.append(skill)
        break

  matched_skills = sorted(
      set(resume_skills) - set(job_skills)
  )

  missing_skills = sorted(
      set(job_skills) - set(resume_skills)
  )

  additional_skills = sorted(
      set(resume_skills) - set(job_skills)
  )

  if len(job_skills) > 0:
    skill_match_score = (
        len(matched_skills) / len(job_skills)
    ) * 100

  else:
    skill_match_score = 0

  documents = [resume_text, job_description]

  vectorizer = TfidfVectorizer(
      stop_words = "english"
  )

  tfidf_matrix = vectorizer.fit_transform(documents)

  similarity_score = cosine_similarity(
      tfidf_matrix[0:1],
      tfidf_matrix[1:2]
  )[0][0]

  text_similarity_score = similarity_score * 100

  llm_analysis = analyze_resume_with_llm(
      resume_text,
      job_description,
      matched_skills,
      missing_skills,
      skill_match_score
  )


  return {
      "text_similarity_score" : text_similarity_score,
      "skill_match_score" : skill_match_score,
      "resume_skills" : resume_skills,
      "job_skills" : job_skills,
      "matched_skills" : matched_skills,
      "missing_skills" : missing_skills,
      "additional_skills" : additional_skills,
      "llm_analysis" : llm_analysis

  }
```

```python
results = analyze_resume(
    resume_text,
    job_description
)
```
